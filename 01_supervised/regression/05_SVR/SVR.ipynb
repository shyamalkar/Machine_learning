# SVR 

`What is SVR ? `

`SVR is a supervised machine learning algorithm used for regression, where the target is continuous.
It is the regression version of the idea behind Support Vector Machine (SVM).
The main idea of SVR is:
Instead of trying to make every prediction exactly correct, SVR tries to find a function where most data points lie within an acceptable error margin.`

# Coding

# Import Libraries 

```python
import numpy as np  # NumPy → numerical calculations
import pandas as pd  # Pandas → data manipulation

from sklearn.datasets import fetch_california_housing  # California Housing → built-in regression dataset

from sklearn.model_selection import GridSearchCV  # GridSearchCV → systematically tests different hyperparameter combinations
from sklearn.model_selection import train_test_split  # Splits data into training and testing sets
from sklearn.pipeline import Pipeline  # Connects preprocessing and model into one workflow
from sklearn.preprocessing import StandardScaler  # Standardizes features before SVR

from sklearn.svm import SVR  # SVR → Support Vector Regression algorithm

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score  # Regression evaluation metrics
```

```python
# Load the dataset

housing = fetch_california_housing(as_frame=True)  # Loads California Housing dataset as a Pandas DataFrame

X = housing.data  # X → input features used to make predictions

y = housing.target  # y → target value we want the model to predict

print(X.shape)  # Shows number of samples and features

print(y.shape)  # Shows number of target values
```

```python
# Look at the data
X.head()
```

```python
# Train test aplit 

X_train, X_test, y_train, y_test = train_test_split(
    X, y,                    # X = features, y = target
    test_size=0.2,            # 20% data → testing, 80% data → training
    random_state=42           # Keeps the same split every time we run the code
)

```

```python
# Let's use standardscaler
# This is very important for SVR . 

# SVR relies on distance and mathematical relationships between features values. 
# our features have very different scales . 

# For Example :- 
# Population → thousands
# Latitude   → around 30–40
# AveRooms   → around 1–10
# MedInc     → around 0–15

# Without scaling large-valued features can have too much influence.

# Create SVR Pipeline 
# We'll start with the RBF kernel, because it is one of the most important SVR configurations for nonlinear relationships.


svr_pipeline = Pipeline([
    ("scaler", StandardScaler()),  # Standardizes every feature so SVR treats features on comparable scales
    ("svr", SVR(kernel="rbf"))      # RBF-SVR → learns nonlinear relationships between features and target
])

# I put standard scaler inside the pipeline
# This is the correct approach because the scaler will be fitted only using the training data during .fit().
```

```python
svr_pipeline.fit(
    X_train,  # Training features
    y_train   # Training target values
)
```

```python
# Make Predictions .

y_pred = svr_pipeline.predict(
    X_test  # Test features used to generate predictions
)
```

```python
# Evaluate SVR
# We will use three important regression matrices . 

rmse = np.sqrt(mean_squared_error(
    y_test,  # Actual target values
    y_pred   # SVR predicted values
))

mae = mean_absolute_error(
    y_test,  # Actual target values
    y_pred   # SVR predicted values
)

r2 = r2_score(
    y_test,  # Actual target values
    y_pred   # SVR predicted values
)

# Print them 


print("RMSE:", rmse)  # RMSE → average prediction error, with larger errors penalized more

print("MAE:", mae)  # MAE → average absolute prediction error

print("R²:", r2)  # R² → proportion of target variance explained by the model
```

` Now the Important part - SVR Hyperparameters . `

` The first Model was just the baseline SVR`

`The real powerr of SVR comes from tuning: `
`C, epsilon, gamma, kernel`


```python

# Low C - Control how strongly SVR penalizes errors outside the  ε-tube.(epsilon)
# → allows more errors
# → simpler model
# → stronger regularization

# High C
# penalizes errors more
# → tries harder to fit training data
# → can overfit





#epsilon - control the size of the epsilon-tube

# Large epsilon
# → more errors are ignored
# → fewer support vectors

# Small epsilon
# → stricter predictions
# → potentially more support vectors





# gamma 
# Especially important for the RBF kernel . 

# Low gamma
# → smoother / broader influence

# High gamma
# individual points have stronger influence
# more complex model
# → possible overfitting








# Define the parameter combinations 

param_grid = {
    "svr__C": [1, 10, 100],  # Tests different C values; "svr__" refers to the SVR step inside the Pipeline
    "svr__epsilon": [0.01, 0.1, 0.2],  # Tests different epsilon-tube widths
    "svr__gamma": ["scale", 0.01, 0.1]  # Tests different RBF influence strengths
}
```

```python
# create GridSearchCV

grid_search = GridSearchCV(
    estimator=svr_pipeline,  # The complete pipeline that GridSearchCV will train and evaluate
    param_grid=param_grid,  # The hyperparameter combinations GridSearchCV will test
    cv=5,  # 5-fold cross-validation → training data is divided into 5 parts and each part becomes validation once
    scoring="neg_root_mean_squared_error",  # Uses RMSE for evaluation; sklearn negates it because higher scores are preferred
    n_jobs=-1  # Uses all available CPU cores to speed up the search
)
# cv=5 is a common practical choice it gives a good balance between: , computional cost, reliable alidation, training data availability . 

# cv=10 is possible, but it approximately doubles the number of model fits compared with cv=5
```

```python
# Run hyperperameter search 

grid_search.fit( # it can takes lot's of time .
    X_train,  # Training features used during cross-validation
    y_train   # Training target values used during cross-validation
)
```

```python
# Find the best parameters . 

print("Best Parameters:", grid_search.best_params_)  # Shows the parameter combination that produced the best CV score

print("Best CV Score:", grid_search.best_score_)  # Shows the best cross-validation score
```

```python
# I can also get the best trained model :

best_svr = grid_search.best_estimator_  # Retrieves the complete best Pipeline found by GridSearchCV
```

```python
# Final prediction 

y_pred_best = best_svr.predict(
    X_test  # Unseen test features used for final evaluation
)
```

```python
# Evaluate 
final_rmse = np.sqrt(mean_squared_error(
    y_test,  # Actual test values
    y_pred_best  # Predictions from the tuned SVR
))

final_mae = mean_absolute_error(
    y_test,  # Actual test values
    y_pred_best  # Predictions from the tuned SVR
)

final_r2 = r2_score(
    y_test,  # Actual test values
    y_pred_best  # Predictions from the tuned SVR
)
```

```python
print("Final RMSE:", final_rmse)  # Lower → better prediction error

print("Final MAE:", final_mae)  # Lower → better average absolute error

print("Final R²:", final_r2)  # Higher → better explained variance
```
