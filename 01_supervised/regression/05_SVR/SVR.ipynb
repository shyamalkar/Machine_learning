# SVR 

`What is SVR ? `

`SVR is a supervised machine learning algorithm used for regression, where the target is continuous.
It is the regression version of the idea behind Support Vector Machine (SVM).
The main idea of SVR is:
Instead of trying to make every prediction exactly correct, SVR tries to find a function where most data points lie within an acceptable error margin.`
