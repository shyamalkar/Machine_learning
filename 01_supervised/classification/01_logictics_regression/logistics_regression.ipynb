`Separating data and labels `

```python
# This code is doing one of the most important steps in a machine-learning project: separating features (input) and labels (output).

X = sonar_data.drop(columns=60) # That means , remove columns 60 from the dataset, so you are removing the M/R column
y = sonar_data[60] # This selects only columns 60 . This is the target level that model need to predict. 

print(X) # is the input data 
```

```python
print(y) # y is the label variable where we want to predict the answer .
```

```python
# Training and Test Data 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, stratify=y, random_state=1)

# X_train The feature data that will be used for train the ML model.
# Suppose i have 207 samples and use 10% for testing . 
# approximately , 207 samples, 186 samples -> X_train, 21 samples -> X_test . 
# so X_train contains the 60 sonar measurements for the trainig samples.

# x_test means contains the feature data that the model hasn't seen during training .
# y train contains the correct answer corresponding to X_train.
# X_train sample       y_train
#     ↓                  ↓
#60 sonar values  →     M
#60 sonar values  →     R
#60 sonar values  →     R
#60 sonar values  →     M

# train_test_split() this is a function from sklearn, and randomly divide your dataset into training data and testing data.
# X contains your feature / input data. 
# y contains the target / label 
# predict label is M and R , and X - sonar measurements and y contains Mine and Rock. 

# test_size = 0.1 , this is particularly important . 0.1 means 10% of the data will be used for testing. and the 90% used for training . that means total 208 and 10% out that means 186 training nad 21 for testing . 

# stratify=y this is another very important parameter, especially for your rock / mine classification
# stratif=y means keep approximately the same proportion of M and R both training and testing datasets.
# if suppose your original dataset looks like: M=111, R=97, The dataset contains roughly , M = 53% , R = 47% . 
# without stratification, random splitting could accidently produce something like: 
# Training: M = 100, R = 86, Testing: M = 11 , R = 11

# It might be okay, but random splitting doesn't guarantee the proportions.
# with stratify=y it looks like =  Original = M ≈ 53% ,R ≈ 47% , - > Training ,M ≈ 53% ,R ≈ 47% , -> Testing M ≈ 53%, R ≈ 47%

# That is why stratify=y is commonly usefull for classification ptobelem. 

# random_state = anynumber,  this controls the randomness of the split . random_state= anynumber means , it fixes the random split so you get the same result every time you run the code.  so everytime you run it, you get the same random split . 
# and there is nothing about sepcial about 42 , you could use , any number . 

# so random_state = any_number a seed that makes random results reproducible.
```

```python
print("Original Data shape",X.shape)
print("X_train_shape", X_train.shape)
print("X_test_shape", X_test.shape)
```

```python
print("Training data",X_train) # it is a trianing data and we see that how the data looks like ?
```

```python
print("Label data",y_train) # y_train is a label data
```

```python
model = LogisticRegression() # we store the algorithm in model variable.

# You can also use SVM , SVM also give good prediction. 
```

```python
# training the Logistics Regression model with training data
model.fit(X_train, y_train) # fit means , train / learn from the data .  model.fit() is the learning stage.

# so i give the model X_train - sonar data and y_train - correct answer (R/M)
# The model try to learn the relationship sonar measuremets -> rock or mine. 

# used for = Because before making predictions on new data, the model needs to learn from your training data.

```

# Model evaluation

```python
# predicting the training data 

X_train_prediction = model.predict(X_train) # "Now that you've learned from X_train, predict the labels of X_train." the model produce something like, 'R', 'M'. , these prediction are stored in: X_train_prediction
# to see how well the model performance on the training data. 

# calculating the training accuracy , 
training_data_accuracy = accuracy_score(X_train_prediction, y_train) # accuracy_score() compares: model_prediction vs actual answer .  and the result store in training_data_accuracy . 
#for example

#Prediction    Actual
#   R             R      ✓
#   M             M      ✓
#   R             M      ✗
#   M             M      ✓

# suppose 180 out of 186 predictions are correct:
# accuracy = 180/186 
#        = 96.8% and this result store in training_data_accuracy
# use for to measure - To measure how accurately your model predicts the training data. 


# print the training accuracy 
```

```python
print("Accuracy on training data:", training_data_accuracy) # this simply displays the result : 
# acuracy on training data 0.83 means 83% 
```

`predict the test data `

```python
# now comes the important part 
# accuracy on test data 

X_test_prediction = model.predict(X_test) # give the model X_test, which contains data that it did't use during training .
test_data_accuracy = accuracy_score(X_test_prediction, y_test)


#This tells us how well the model performs on unseen data. This is much more useful for judging whether your model can generalize . 

# predicting values on actual values . 

# Prediction    Actual
# R             R      ✓
# M             M      ✓
# M             R      ✗
# R             R      ✓

# then we calculate the percentage of correct predictions. 
```

```python
print("Äccuracy on test data", test_data_accuracy) # print the test accuracy . 
# if accuracy on test data: 0.7619 that means 76.19% accuracy. 
```
