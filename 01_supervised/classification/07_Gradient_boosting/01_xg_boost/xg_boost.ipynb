# Import libraries

```python
import numpy as np                         # Numerical operations
import pandas as pd                        # Data handling

from sklearn.datasets import load_breast_cancer  # Load prepared dataset

from sklearn.model_selection import train_test_split  # Split data

from sklearn.metrics import (
    accuracy_score,                         # Calculate accuracy
    confusion_matrix,                      # Create confusion matrix
    classification_report                  # Precision, recall and F1
)

from xgboost import XGBClassifier          # XGBoost classification algorithm
```

# Load the dataset

```python
data = load_breast_cancer()  # Load the Breast Cancer dataset
```

# Understnad the dataset

```python
print(data.data.shape)  # Shows number of samples and features
```

# Check features names

```python
print(data.feature_names)  # Shows all 30 feature names
```

# Check target 

```python
print(data.target.shape)       # Shows number of target labels
print(data.target_names)       # Shows names of the classes
```

# check class distribution 

```python
print(
    pd.Series(data.target).value_counts()
)  # Shows how many samples belong to each class
```

# Separate X and y 

```python
X = data.data      # X contains the 30 input features
y = data.target    # y contains the target labels: 0 or 1
```

```python
print("X shape:", X.shape)  # Should be (569, 30)
print("y shape:", y.shape)  # Should be (569,)
```

print("X shape:", X.shape)  # Should be (569, 30)
print("y shape:", y.shape)  # Should be (569,)


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,                  # Input features
    y,                  # Target labels
    test_size=0.2,      # 20% data is reserved for testing
    stratify=y,         # Keeps class proportions similar
    random_state=42     # Makes the split reproducible
)
```

# Create XGBoost model

```python
classifier = XGBClassifier(
    random_state=42,     # Makes results reproducible
    eval_metric="logloss" # Evaluation metric for binary classification
)
```

# Train the model

```python
classifier.fit(
    X_train,   # Training features
    y_train    # Training labels
)
```

# Training prediction

```python
X_train_prediction = classifier.predict(
    X_train
)  # Predict labels for training data

# Calculate accuracy 

training_data_accuracy = accuracy_score(
    y_train,              # Actual training labels
    X_train_prediction    # Predicted training labels
)

print(
    "Training Accuracy:",
    training_data_accuracy
)
```

# Test prediction

```python
X_test_prediction = classifier.predict(
    X_test
)  # Predict labels for unseen test data
```

# Test accuracy 

```python
test_data_accuracy = accuracy_score(
    y_test,             # Actual test labels
    X_test_prediction   # Predicted test labels
)

print(
    "Test Accuracy:",
    test_data_accuracy
)
```

# Calculate accuracy gap

```python
accuracy_gap = (
    training_data_accuracy
    - test_data_accuracy
)  # Difference between training and testing accuracy

print(
    "Accuracy Gap:",
    accuracy_gap
)
```

# Confusion matrix 

```python
conf_matrix = confusion_matrix(
    y_test,               # Actual labels
    X_test_prediction     # Predicted labels
)

print("Confusion Matrix:")
print(conf_matrix)
```

# Classification report 

```python
print(
    classification_report(
        y_test,
        X_test_prediction,
        target_names=data.target_names
    )
)

# Not bad performing our dataset
```

# Let's Experiment with XGBoost parameters . 

```python
n_estimators_values = [
    30,
    50,
    100,
    150,
    200
]

results = []

for n in n_estimators_values:

    model = XGBClassifier(
        n_estimators=n,       # Number of boosting trees
        random_state=42,      # Reproducibility
        eval_metric="logloss" # Evaluation metric
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "n_estimators": n,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })

n_estimators_results = pd.DataFrame(results)

print(n_estimators_results)

```

# Experiment learning_rate=number , 

`This controls how strongly each new tree contributes.`

```python
learning_rate_values = [
    0.01,
    0.05,
    0.1,
    0.2
]

results = []

for lr in learning_rate_values:

    model = XGBClassifier(
        n_estimators=100,      # Keep trees fixed
        learning_rate=lr,      # Change learning rate
        random_state=42,
        eval_metric="logloss"
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "Learning Rate": lr,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })

learning_rate_results = pd.DataFrame(results)

print(learning_rate_results)
```

# Experiment with max_depth

```python
max_depth_values = [
    1,
    2,
    3,
    4,
    5
]

results = []

for depth in max_depth_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=depth,       # Change tree depth
        random_state=42,
        eval_metric="logloss"
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "Max Depth": depth,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })

max_depth_results = pd.DataFrame(results)

print(max_depth_results)
```

# Experiment with min_child_weight


`This control how much minimum weight a chield node hsould have before a tree split is allowed . `

```python
min_child_weight_values = [
    1,
    3,
    5,
    10
]

results = []

for weight in min_child_weight_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        min_child_weight=weight,
        random_state=42,
        eval_metric="logloss"
    )

    model.fit(X_train, y_train)

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "min_child_weight": weight,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })

min_child_results = pd.DataFrame(results)

print(min_child_results)
```

# Experiment with subsample

`subsample controls what fraction of training samples are used for each boosting stage.`

```python
subsample_values = [
    0.6,
    0.8,
    1.0
]
results = []

for sample in subsample_values:

    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        subsample=sample,
        random_state=42,
        eval_metric="logloss"
    )

    model.fit(X_train, y_train)

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "Subsample": sample,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })

subsample_results = pd.DataFrame(results)

print(subsample_results)
```

# Create final XGBoost Model

`Compare  experiment , and remember don't pickup high train accuracy `

`but the most simple rule is that - test accuracy - accuracy gap - good candidate`

`For example :- `

`Model A`
`Train  = 100% , Test = 95%, Gap=5%`

`Model B`
`Train = 98%`
`Test = 95%`
`Gap = 3%`

`Both have the same test accuracy, but Model B may be preferable because it has a smaller gap.

But remember: a low gap by itself isn't automatically better.`


`After you run all experiments, choose the values that performed well on your results.`

```python
final_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    min_child_weight=1,
    subsample=0.8,
    random_state=42,
    eval_metric="logloss"
)
```

# Train the final model 

```python
final_model.fit(
    X_train,
    y_train
)
```

# Final Evaluation 

```python
final_prediction = final_model.predict(
    X_test
)

final_accuracy = accuracy_score(
    y_test,
    final_prediction
)

print(
    "Final Test Accuracy:",
    final_accuracy
)
```

# Final Confusion matrix

```python
print("Final Confusion Matrix:")

print(
    confusion_matrix(
        y_test,
        final_prediction
    )
)
```

# Final Classification report 

```python
print("Final Classification Report:")

print(
    classification_report(
        y_test,
        final_prediction,
        target_names=data.target_names
    )
)
```

# Save the final XGBoost Model 

```python
import pickle

# Save the trained XGBoost model
with open(
    "xgboost_breast_cancer.pkl",
    "wb"
) as file:

    pickle.dump(
        final_model,
        file
    )

print("XGBoost model saved successfully!")
```
