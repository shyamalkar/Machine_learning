# `Gradient Boosting`

```python
# What is Gradient Boosting ? 
# Gradient Boosting is an ensemble machine learning technique that builds decision trees sequentially, where each new tree tries to correct the errors made by the previous trees.


# Why do we need Gradient Boosting ? 

# A single decision tree not be powerfull enough . 
# decision tree - prediction - some mistake.  so instead of creating one giant tree ,  gradient boosting says. Let's create a sequence small trees, where each new tree improves the current model .


# So: 
# Weak Tree
#    ↓
# Improvement
#    ↓
# Another Weak Tree
#    ↓
# Improvement
#    ↓
# Another Weak Tree
#    ↓
# ...
#    ↓
# Strong Ensemble
# This is the idea of Boosting . 





# What is boosting ? 
# Easy defination - Boosting combines multiple weak learners sequentially to create a stronger predictive model. 


# But How does the Gradient Boosting work ? 
# Suppose we are pedicting house prices . 
# Actual price = 100, 200, 300. 
# Our First model makes: 
# 80, 220, 250 . 
# errors: actual prediction 
# 100 - 80 = +20 
#
#
```

# GRadient Boosting 

Gradient boosting is a powerful supervised machine learning technique that builds an ensemble of weak prediction models—typically small decision trees—sequentially to minimize prediction errors

# Import libraries

```python
import numpy as np  # Import numpy for numerical operation 
import pandas as pd # improt pandas ffor tabular data

from sklearn.datasets import load_breast_cancer # load the breast cancer
from sklearn.model_selection import train_test_split # train and test and split 
from sklearn.ensemble import GradientBoostingClassifier # my machine learning algorithm

from sklearn.metrics import (accuracy_score, # predict how many predictions by class
                             
                             confusion_matrix, # confusion_matrix shows correct and incorrect predictions by class
                               classification_report) # classification precision, recall and F1-Score
```

```python
data = load_breast_cancer()
```

```python
# check feature names 

data.feature_names
```

```python
# check target 
print("Target Shape:", data.target.shape)
# 569 is  sample
```

```python
# target names 
print("Target names",data.target_names)
```

`Check class distribution`

```python
print("Class Distribution:")
print(pd.Series(data.target).value_counts())
```

`Separate X and y`

```python
X = data.data
y = data.target

print("Input features:-\n", X)
```

```python
print("Target features :- \n", y)
```

# Train and test and Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, stratify=y, random_state=42)
```

```python
classifier = GradientBoostingClassifier(random_state=42)
```

# But what exactly do Gradient boositng classifier ? 

`In Random forest `

`builds many individual decision trees, collects all of their individual predictions, and combines them into one final result.`

`But in Gradient Boosting `
`Gradient boosting builds many weak decision trees sequentially`

`Think like :- ` , `Tree 1 - Makes predictions - Finds errors - Tree 2 focuses on errors - Finds remaining errors - Tree 3 focuses on remaining errors - ... - Final combined model `

`The trees are built sequentially, with later trees trying to improve the mistakes of earlier ones.`

# Train the model

```python
classifier.fit(X_train, y_train)
# now the  gradient boosting learns patterns from the training data . 
```

# Train Prediction

```python
X_train_prediction = classifier.predict(X_train)


# calculate the accuracy 

train_data_accuracy = accuracy_score(y_train, X_train_prediction)
print("Train Accuracy:", train_data_accuracy)
```

# Test Prediction

```python
X_test_prediction = classifier.predict(X_test)

# Calculate the accuracy 

test_data_accuracy = accuracy_score(y_test, X_test_prediction)

print("Test Accuracy", test_data_accuracy)
```

`Calculate the train and test accuracy `

```python
gap = train_data_accuracy - test_data_accuracy
print("Train and test accuracy gap",gap)
print("Train Data accuracy", train_data_accuracy)
print("Test Data accuracy", test_data_accuracy)
```

# Confusion matrix

```python
conf_matrix = confusion_matrix(
    y_test,
    X_test_prediction
)

print("Confusion Matrix:")
print(conf_matrix)
```

# Classification report

```python
report = classification_report(
    y_test,
    X_test_prediction,
    target_names=data.target_names
)

print(report)
```

# Gradient Boosting Experiment

# Experiment with n_estimator=number

`How many boosting stages/trees should the model build ?`



```python
n_estimators_values = [
    30,
    40,
    50,
    100,
    150,
    200
]
results = []

for n in n_estimators_values:

    model = GradientBoostingClassifier(
        n_estimators=n,
        random_state=42
    )

    model.fit(X_train,y_train)

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "n_estimators": n,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


n_estimators_results = pd.DataFrame(results)

print(n_estimators_results)

print("What are we learning here ? ")
print("more trees - more opportunities to improve the model but too many trees - more complexity - possible overfitting")
```

# Experiment learning_reate = number

`This is one of the most important gradient boosting parameters.`

`It control how strongly each new tree contributes to the final model.`

```python
learning_rate_values = [
    0.01,
    0.05,
    0.1,
    0.2
]
results = []

for learning_rate in learning_rate_values:

    model = GradientBoostingClassifier(
        learning_rate=learning_rate,
        n_estimators=100,
        random_state=42
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "Learning Rate": learning_rate,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


learning_rate_results = pd.DataFrame(results)

print(learning_rate_results)
```

# Experiment with max_depth=number

`Gradient Boosting uses decision trees as its weak learners.`

```python
# max_depth controls how deep those trees can become.

max_depth_values = [1, 2, 3, 4, 5]

results = []

for depth in max_depth_values:

    model = GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=depth,
        random_state=42
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(X_train)
    test_prediction = model.predict(X_test)

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "Max Depth": depth,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


max_depth_results = pd.DataFrame(results)

print(max_depth_results)

# Always remember 

#Depth too small
#      ↓
#Trees too weak
#      ↓
#Possible underfitting


#Depth balanced
#      ↓
# Good learning


# Depth too large
#      ↓
# Complex trees
#      ↓
# Possible overfitting
```

# Feature importance 

```python
feature_importance = pd.DataFrame({
    "Feature": data.feature_names,
    "Importance": classifier.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("This tells us which features contributed more to the model's tree based decisions:-.\n")
print(feature_importance)

```

# Select the final model 


```python
final_model = GradientBoostingClassifier(
    n_estimators=100,       #  n_estimators experiment
    learning_rate=0.10,    # learning-rate experiment
    max_depth=1,            # max-depth experiment
    random_state=42
)
```

```python
final_model.fit(X_train, y_train)

train_prediction = final_model.predict(X_train)
test_prediction = final_model.predict(X_test)

train_accuracy = accuracy_score(
    y_train,
    train_prediction
)

test_accuracy = accuracy_score(
    y_test,
    test_prediction
)

print("Final Training Accuracy:", train_accuracy)
print("Final Test Accuracy:", test_accuracy)
print(
    "Accuracy Gap:",
    train_accuracy - test_accuracy
)


print("Confusion Matrix:")
print(
    confusion_matrix(
        y_test,
        test_prediction
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        test_prediction,
        target_names=data.target_names
    )
)
```
