# `Gradient Boosting`

```python
# What is Gradient Boosting ? 
# Gradient Boosting is an ensemble machine learning technique that builds decision trees sequentially, where each new tree tries to correct the errors made by the previous trees.


# Why do we need Gradient Boosting ? 

# A single decision tree not be powerfull enough . 
# decision tree - prediction - some mistake.  so instead of creating one giant tree ,  gradient boosting says. Let's create a sequence small trees, where each new tree improves the current model .


# So: 
# Weak Tree
#    ↓
# Improvement
#    ↓
# Another Weak Tree
#    ↓
# Improvement
#    ↓
# Another Weak Tree
#    ↓
# ...
#    ↓
# Strong Ensemble
# This is the idea of Boosting . 





# What is boosting ? 
# Easy defination - Boosting combines multiple weak learners sequentially to create a stronger predictive model. 


# But How does the Gradient Boosting work ? 
# Suppose we are pedicting house prices . 
# Actual price = 100, 200, 300. 
# Our First model makes: 
# 80, 220, 250 . 
# errors: actual prediction 
# 100 - 80 = +20 
#
#
```
