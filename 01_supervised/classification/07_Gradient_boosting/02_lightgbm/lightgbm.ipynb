```python
import numpy as np                         # Numerical operations
import pandas as pd                        # Data handling

from sklearn.datasets import load_breast_cancer
# Loads the prepared Breast Cancer dataset

from sklearn.model_selection import train_test_split
# Splits the dataset into training and testing data

from sklearn.metrics import (
    accuracy_score,                        # Calculates accuracy
    confusion_matrix,                      # Creates confusion matrix
    classification_report                  # Precision, recall and F1
)

from lightgbm import LGBMClassifier
# LightGBM classification algorithm
```

# Load the dataset

```python
data = load_breast_cancer()
```

# Understnad the dataset

```python
print("Dataset Shape:", data.data.shape)
# Shows number of samples and number of features
print("First one is sampel and second one is features")
```

# Check feature names

```python
print("Feature Names:")
print(data.feature_names)
# Displays the names of the 30 features
```

# Target names

```python
print("Target Shape:", data.target.shape)
# Shows how many target labels exist

print("Target Names:")
print(data.target_names)
# Shows the meaning of target values
```

# Check class distribution 

```python
print("Target Shape:", data.target.shape)
# Shows how many target labels exist

print("Target Names:")
print(data.target_names)
# Shows the meaning of target values
```

# Separate X and y 

```python
X = data.data
# X contains the 30 input features 

y = data.target
# y contains the target labels 
```

```python
print("Input features X Shape:", X.shape)
print("Target features y Shape:", y.shape)
```

 # Train test split 

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,                  # Input features
    y,                  # Target labels
    test_size=0.2,      # 20% data for testing
    stratify=y,         # Keeps class proportions similar
    random_state=42     # Makes the split reproducible
)
```

# Create Baseline LightGBM Model


```python
classifier = LGBMClassifier(
    random_state=42, # make it reproducible 
    verbosity=-1 # reduces LightGBM's Extra training messages
)
# Creates the LightGBM classification model
```

# Train the model 

```python
classifier.fit(X_train, y_train)

# Train LigthGBM using the training data . 

# Now the model learned patterns from the training data. 
```

# Training prediction 


```python
X_train_prediction = classifier.predict(
    X_train
)
# Predicts classes for the training data

# Accuracy Calculate 

training_data_accuracy = accuracy_score(
    y_train,
    X_train_prediction
)
# Compares actual training labels with predicted labels

print(
    "Training Accuracy:",
    training_data_accuracy
)
```

# Test Prediction 

```python
X_test_prediction = classifier.predict(X_test)

# predict classes for unsen test data 

# Calculate accuracy 

test_data_accuracy = accuracy_score(
    y_test,
    X_test_prediction
)
# Compares actual test labels with predictions

print(
    "Test Accuracy:",
    test_data_accuracy
)
```

# Accuracy gap 

```python
accuracy_gap = (
    training_data_accuracy
    - test_data_accuracy
)
# Measures the difference between training and test accuracy

print(
    "Accuracy Gap:",
    accuracy_gap
)
```

# Confusion matrix 

```python
conf_matrix = confusion_matrix(
    y_test,
    X_test_prediction
)
# Creates a matrix showing correct and incorrect predictions

print("Confusion Matrix:")
print(conf_matrix)
```

# Classification report 


```python
print(
    classification_report(
        y_test,
        X_test_prediction,
        target_names=data.target_names
    )
)
# Shows precision, recall, F1-score and support
```

# Experiment  - 
`n_estimators - means the number of boosting iterations/trees.`

```python
n_estimators_values = [
    30,
    50,
    100,
    150,
    200
]

results = []

for n in n_estimators_values:

    model = LGBMClassifier(
        n_estimators=n,
        random_state=42,
        verbosity=-1
    )

    # Train the model
    model.fit(
        X_train,
        y_train
    )

    # Training prediction
    train_prediction = model.predict(
        X_train
    )

    # Test prediction
    test_prediction = model.predict(
        X_test
    )

    # Training accuracy
    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    # Test accuracy
    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    # Difference between training and testing
    gap = train_accuracy - test_accuracy

    results.append({
        "n_estimators": n,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


n_estimators_results = pd.DataFrame(results)

print(n_estimators_results)
```

# Experiment with learning rate 

`This is a very important LightGBM parameter. `

```python
learning_rate_values = [
    0.01,
    0.05,
    0.1,
    0.2
]


results = []

for lr in learning_rate_values:

    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=lr,
        random_state=42,
        verbosity=-1
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(
        X_train
    )

    test_prediction = model.predict(
        X_test
    )

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "Learning Rate": lr,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


learning_rate_results = pd.DataFrame(results)

print(learning_rate_results)
```

# Experiment - num_leaves


`This is very importatn LightGBM parameter.`

`Unlike the gradient Boosting Experiment i did earlier, LightGBM's num_leaves is particularly important because LightGBM uses leaf-wise tree growth`

`num_leaves controls the maximum number of leaves in a tree.`

```python
num_leaves_values = [
    7,
    15,
    31,
    50,
    75
]



results = []

for leaves in num_leaves_values:

    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        num_leaves=leaves,
        random_state=42,
        verbosity=-1
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(
        X_train
    )

    test_prediction = model.predict(
        X_test
    )

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "num_leaves": leaves,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


num_leaves_results = pd.DataFrame(results)

print(num_leaves_results)
```

# Experiment max_depth
`max_depth limits how deep the tree can grow.`

```python
max_depth_values = [
    2,
    3,
    4,
    5,
    7
]


results = []

for depth in max_depth_values:

    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        num_leaves=31,
        max_depth=depth,
        random_state=42,
        verbosity=-1
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(
        X_train
    )

    test_prediction = model.predict(
        X_test
    )

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "Max Depth": depth,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


max_depth_results = pd.DataFrame(results)

print(max_depth_results)
```

# Experiment min_child_samples

`This controls the minimum number of samples required in a leaf.`

```python
min_child_samples_values = [
    5,
    10,
    20,
    30,
    50
]

results = []

for samples in min_child_samples_values:

    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        num_leaves=31,
        min_child_samples=samples,
        random_state=42,
        verbosity=-1
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(
        X_train
    )

    test_prediction = model.predict(
        X_test
    )

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "min_child_samples": samples,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


min_child_results = pd.DataFrame(results)

print(min_child_results)
```

# Experiment - colsamples_bytree

`This controls the fraction of features used when building each tree.`

```python
colsample_values = [
    0.6,
    0.8,
    1.0
]

results = []

for fraction in colsample_values:

    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        num_leaves=31,
        colsample_bytree=fraction,
        random_state=42,
        verbosity=-1
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(
        X_train
    )

    test_prediction = model.predict(
        X_test
    )

    train_accuracy = accuracy_score(
        y_train,
        train_prediction
    )

    test_accuracy = accuracy_score(
        y_test,
        test_prediction
    )

    gap = train_accuracy - test_accuracy

    results.append({
        "colsample_bytree": fraction,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Accuracy Gap": gap
    })


colsample_results = pd.DataFrame(results)

print(colsample_results)
```
