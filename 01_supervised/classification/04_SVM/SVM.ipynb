```python
# What is SVM ? 
# Support Vector Machine (SVM) is a supervised machine learning algorithm that finds the best decision boundary to separate different classes while trying to maximize the margin between the classes.

# SVM can be used for :- 

#Classification and regression . 

# suppose we have 2 classes . 
# there could be many possible boundaries

# What is Margin ? 
# The margin is the distance between the decision boundary and the closest training points from each class.
# this is why SVM is called a maximum margin classifier. 
```
