# `Naive_bayes`

```python
# What is Naive Bayes ? 
# Naive Bayes is a supervised classification algorithm based on Bayes' theorem that calculates the probability of each possible class and predicts the class with the highest probability.

#Why do we need naive bayes ? 
# Supose we want to classify an email:
# congratulations ! you won a free prize!
# i want to predict - spam or not spam 
# naive bayes looks at evidence such as:
# "free"       → Spam?
# "prize"      → Spam?
# "won"        → Spam?
# "congratulations" → Spam?
# i combines these of evidence to calculate :
# p (Spam | words)
# p (Not Spam | words)
# then: higher probability and final prediction . 

# So the basic idea is = Calculate the probability of each class given the observed features, then choose the class with the highest probability.


# Types of Naive Bayes 
# This is important becuase different types are suited to different kinds of data . 

"""Gaussian Naive Bayes - gaussianNb() , used when features are continuous and are reasonably modeled by gaussian distribution - age, height, weight, temparature , blood pressure."""

# Multinomial Naive bayes - MultinomialNB() , very commonly used for: - Text classification, document classification, spam detection, word counts . nernouliNB - used when - 0/1 , True and False , Prsesent/absent. categorical naive bayes - used for categorical naive bayes , used for  - color, red, blue, green, weather - sunny , rainy, cloudly
```

# Coding

# Import libraries

```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split

from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

# Load the iris dataset

```python
data = load_iris()
```

`Check the dataset shape`

```python
print("Data Shape :", data.data.shape)
# first number is sample and second number is features. 
```

`What are the 4 features ?`

```python
print("Features name:")
print(data.feature_names)
```

`Check target`

```python
print("Target Shape:", data.target.shape)
 # output would be 150 , what does mean 150 ? 150 means 150 flowers. 
```

`Target Values`

```python
data.target
# total 3 class

# 0 stand for setosa 
# 1 stand for versicolor
# 2 stand for virginica 

# this is a multiclass classification problem . 
```

`Check target names`

```python
data.target_names
```

`Separate X and y `

```python
X = data.data 
y = data.target

print("Input features:", X)
```

```python
print("Label features", y)
```

` Traiin Test Split `

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, # 20% data for testing and other 80% dataset for training .
                                                     stratify=y, # preserve class proportion
                                                     random_state = 42) # reprodusible split 
```

`Create gaussian naive bayes`

```python
classifier = GaussianNB()
```

```python
classifier.fit(X_train, y_train)
# During training, GaussianNB estimates information about each feature for each class.
```

`Train prediction`

```python
X_train_prediction = classifier.predict(X_train)

# accuracy score

X_train_accuracy = accuracy_score(y_train, X_train_prediction)
print("Test accuracy:", X_train_accuracy)
```

`Test prediction`

```python
X_test_prediction = classifier.predict(X_test)

# accuracy score

test_data_accuracy = accuracy_score(y_test, X_test_prediction)
print("Test Data accuracy", test_data_accuracy
      )
```

# Try to know Overfit or Underfit 

```python
accuracy_gap = X_train_accuracy - test_data_accuracy 
print("X_train_prediction:- ", X_train_accuracy)
print("test_data_accuracy:- ", test_data_accuracy)
print("Accurayc gap:-", accuracy_gap)

# Not a huge gap model performing good at testing  i mean unseen dataset . 
```

# Confusion Matrix

```python
# Create confusion matrix using actual and predicted test labels
conf_matrix = confusion_matrix(
    y_test,                 # Actual labels
    X_test_prediction       # Model's predicted labels
)

print("Confusion Matrix, 3 * 3 because iris has 3 classes ")
print(conf_matrix)
```

# classification report 

```python


# Generate precision, recall, F1-score and support
report = classification_report(
    y_test,
    X_test_prediction,
    target_names=data.target_names
)

print(report)
```

# Save the first Naive Bayes 


```python
import pickle 

with open ("Naive_bayes_train.pkl", 'wb') as file:
    pickle.dump(classifier,file)

print("Print the naove bayes train successfully ")
```
