# Dataset 

`For this topic, a sequence dataset is more qppropriate than the supermarket, transaction dataset we used fro FP-Growth/ECLAT`

we will create a small customer-purchase sequence dataset.

```python
sequences = [
    [["Phone"], ["Earphones"], ["Phone Case"]],       # Customer 1
    [["Phone"], ["Earphones"], ["Phone Case"]],       # Customer 2
    [["Phone"], ["Earphones"], ["Charger"]],          # Customer 3
    [["Laptop"], ["Mouse"], ["Keyboard"]],            # Customer 4
    [["Phone"], ["Earphones"], ["Charger"]],          # Customer 5
    [["Phone"], ["Phone Case"], ["Earphones"]],       # Customer 6
    [["Laptop"], ["Mouse"], ["Keyboard"]],            # Customer 7
]

# here

# Customer 1
#    ↓
# Phone
#    ↓
# Earphones
#   ↓
# Phone Case
```

# Why is this different from FP-Growth/ECLAT ?

consider 

Customer A → Phone → Earphones

Customer B → Earphones → Phone

 But both customer have the same 2 items . 
 But sequential mining sees. 

A → Phone → Earphones

B → Earphones → Phone


They are different sequences. 
THat is the central idea. 



`What we are trying to find ?`

# FInd frequent individual items. 


```python
from collections import Counter                         # Count how frequently each item appears


item_counts = Counter()                                # Store number of customers containing each item


for sequence in sequences:                             # Loop through each customer's sequence

    for item in sequence:                              # Loop through items in the sequence

        item_counts[item] += 1                         # Count the item


print(item_counts)
```

# Calculate support 

```python
min_support = 0.4                                     # Require a pattern to appear in at least 40% of customers

n_sequences = len(sequences)                           # Total number of customer sequences


for item, count in item_counts.items():

    support = count / n_sequences                      # Calculate support of the item

    if support >= min_support:

        print(
            item,
            "→ support:",
            round(support, 2)
        )
```

# FInd the sequential pairs. 


```python
# We need to check whether A occur before B.  


from itertools import permutations                        # Generate possible ordered pairs


items = list(item_counts.keys())                           # Get all unique items


patterns = []                                              # Store frequent sequential patterns


for item1, item2 in permutations(items, 2):                # Generate ordered pairs

    count = 0                                               # Count sequences containing item1 before item2


    for sequence in sequences:                              # Examine every customer sequence

        if (
            item1 in sequence                              # Check whether item1 exists
            and
            item2 in sequence                              # Check whether item2 exists
            and
            sequence.index(item1) < sequence.index(item2)  # Check whether item1 occurs before item2
        ):

            count += 1                                      # Pattern found in this sequence


    support = count / n_sequences                           # Calculate sequential support


    if support >= min_support:

        patterns.append(
            (
                item1,
                item2,
                support
            )
        )


print(patterns)

# For example, if i get 
# Phone - earphones. with relatively highsupport. 
```
