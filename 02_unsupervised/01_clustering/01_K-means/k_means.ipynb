# Coding

```python
# Import all libraries

# NumPy → numerical calculations
import numpy as np

# Pandas → data manipulation
import pandas as pd

# Matplotlib → visualization
import matplotlib.pyplot as plt

# Iris dataset → built-in dataset
from sklearn.datasets import load_iris

# StandardScaler → standardizes features so distance calculations are fair
from sklearn.preprocessing import StandardScaler

# KMeans → K-Means clustering algorithm
from sklearn.cluster import KMeans

# Silhouette Score → evaluates how well-separated the clusters are
from sklearn.metrics import silhouette_score
```

```python
# Load the dataset 

iris = load_iris()                       # Loads the Iris dataset

X = iris.data                            # Features used by K-Means

y = iris.target                          # True species labels, NOT used for training
```

```python
# Check the data 

X.shape
```

```python

iris.feature_names
```

```python
# Feature scaling 
# k-,eans is distance-based. therefore, feature scale can affect the clustering . 


scaler = StandardScaler()                # Creates the standardization object

X_scaled = scaler.fit_transform(
    X                                     # Learns feature statistics and scales the data
)
```

```python
kmeans = KMeans(
    n_clusters=3,                        # Creates 3 clusters
    init="k-means++",                    # Smart centroid initialization
    n_init=10,                           # Runs the algorithm 10 times with different initial centroids
    max_iter=300,                        # Maximum 300 iterations for each run
    random_state=42                      # Makes the result reproducible
)

kmeans.fit(
    X_scaled                             # K-Means learns the clusters from X only
)
```

```python
# Notice something important :
kmeans.fit(X_scaled) # Notice there is no y_train . 
# That's because that is iunsupervised learning . 
```

```python
# Get cluster labels
clusters = kmeans.labels_                # Cluster assigned to every observation

print(clusters[:20])                     # Display the first 20 cluster assignments


# The numbers - 0, 1, 2
```

```python
# Get the centroids

centroids = kmeans.cluster_centers_ # Coordinates of the learned cluster center 

# These are the centers that k-means learned

# Evaluate k-means
# inertia 

inertia = kmeans.inertia_                # Measures total squared distance from points to their centroids

print("Inertia:", inertia)

# Silhouette score
# This is one of the most usefull clustering evaluation metrics . 
# it measure - cohesion - are points close to their own cluster . 
# separation - are different cluster sufficiently far apart ?
# The score ranges approximately from  --     -1 ------0-------- +1

# Calculate it 

silhouette = silhouette_score(
    X_scaled,                            # Feature data used for clustering
    clusters                             # Cluster assignments produced by K-Means
)

print("Silhouette Score:", silhouette) # higher is generally better . 
```

```python
K_values = range(2, 11)                  # Test K from 2 through 10

inertias = []                            # Stores inertia for each K

silhouette_scores = []                   # Stores silhouette score for each K

for k in K_values:

    model = KMeans(
        n_clusters=k,                    # Try this value of K
        init="k-means++",                # Smart centroid initialization
        n_init=10,                       # Multiple initializations
        max_iter=300,                    # Maximum iterations
        random_state=42                  # Reproducibility
    )

    labels = model.fit_predict(
        X_scaled                         # Fit K-Means and return cluster assignments
    )

    inertias.append(
        model.inertia_                   # Save inertia for this K
    )

    silhouette_scores.append(
        silhouette_score(
            X_scaled,                    # Dataset
            labels                       # Cluster assignments
        )
    )


```

```python
# Plot the elbow 

plt.figure(figsize=(8, 5))

plt.plot(
    K_values,                            # Number of clusters
    inertias,                            # Inertia for each K
    marker="o"                           # Show each K value
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    K_values,                            # Number of clusters
    silhouette_scores,                   # Silhouette score for each K
    marker="o"                           # Show each score
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score vs K")

plt.show()
```

```python
# Final K-means model 

final_kmeans = KMeans(
    n_clusters=3,                        # Final number of clusters
    init="k-means++",                    # Smart initialization
    n_init=10,                           # Multiple initializations
    max_iter=300,                        # Maximum iterations
    random_state=42                      # Reproducibility
)

final_clusters = final_kmeans.fit_predict(
    X_scaled                             # Learn clusters and return assignments
)
```

```python
# Final evaluation

final_silhouette = silhouette_score(
    X_scaled,                            # Scaled dataset
    final_clusters                       # Final cluster assignments
)

print("Final Inertia:", final_kmeans.inertia_)

print("Final Silhouette Score:", final_silhouette)
```

```python
# Visualize the cluster 

plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 2],                      # Petal length
    X_scaled[:, 3],                      # Petal width
    c=final_clusters,                    # Color according to cluster
    s=50                                 # Size of points
)

plt.scatter(
    final_kmeans.cluster_centers_[:, 2], # Centroid petal length
    final_kmeans.cluster_centers_[:, 3], # Centroid petal width
    marker="X",                          # Mark centroids with X
    s=200                                # Make centroids larger
)

plt.xlabel("Petal Length (Scaled)")
plt.ylabel("Petal Width (Scaled)")
plt.title("K-Means Clustering")

plt.show()
```
